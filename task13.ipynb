Q-11: Softmax function
Create a Python function to calculate the Softmax of the given numpy 1D array. The function only accepts the numpy 1D array, otherwise raise error.
  
$$
\sigma(\mathbf{z})_i = \frac{e^{z_i}}{\sum_{j=1}^{K} e^{z_j}}
$$  

https://en.wikipedia.org/wiki/Softmax_function  

Example 1:  

Input:  

[86.03331084, 37.7285648,  48.64908087, 87.16563062, 38.40852563, 37.20006318]  

Output:  
[2.43733249e-01, 2.56112115e-22, 1.41628284e-17, 7.56266751e-01, 
       5.05514197e-22, 1.50974911e-22]  


Example 2:  

Input:  
[33.17344305, 45.61961654, 82.05405781, 80.9647098,  68.82830233, 91.52064278]  

Output:  
[4.57181035e-26, 1.16249923e-20, 7.73872596e-05, 2.60358426e-05,
       1.39571531e-10, 9.99896577e-01]  


```python
def softmax(array):
    if type(array)!=np.ndarray:
        raise TypeError("requires Numpy array")
    elif array.ndim !=1:
        raise ValueError(f"Expected a 1d array, but received a {array.ndim}d array")

    s =np.sum(np.exp(array))
    return np.exp(array)/s

softmax(np.array([86.03331084, 37.7285648,  48.64908087, 87.16563062, 38.40852563, 37.20006318]))
softmax(np.array([33.17344305, 45.61961654, 82.05405781, 80.9647098,  68.82830233, 91.52064278]))
# softmax(np.array([[1,2],[3,4]]))
```

Q-13: Dates
Create a python function named date_array that accepts two dates as string format and returns a numpy array of dates between those 2 dates. The function only accept 2 strings, otherwise raise error. The date format should be like this only: 2022-12-6. The end date should be included and for simplicity, choose dates from a same year.  

Example 1:  

Input:  

date_array(start = '2020-09-15', end = '2020-09-25')  
Output:  
  
['2020-09-15', '2020-09-16', '2020-09-17', '2020-09-18',  
 '2020-09-19', '2020-09-20', '2020-09-21', '2020-09-22',  
 '2020-09-23', '2020-09-24', '2020-09-25']  
Example 2:  
  
Input:  

date_array(start = '2022-12-01', end = '2022-12-06')  
Output:  

['2022-12-01', '2022-12-02', '2022-12-03', '2022-12-04', '2022-12-05', '2022-12-06']  
Example 3:  

Input:  
  
date_array(start = '2020-11-25', end = '2020-11-30')  
Output:  

['2020-11-25', '2020-11-26', '2020-11-27', '2020-11-28',  
 '2020-11-29', '2020-11-30']  

  

```python
from datetime import datetime,timedelta
# import datetime
def date_array(start:str,end:str):
    '''date_array that accepts two dates as string format and returns a numpy array of dates between those 2 dates. The function only accept 2 strings, otherwise raise error. The date format should be like this only: 2022-12-6.(yy-mm-dd)'''
    if type(start) != str | type(end)!= str:
        raise TypeError("s tart and end must be strings")
    try:
        # Convert strings to datetime objects
        start_date = datetime.strptime(start, "%Y-%m-%d")
        end_date = datetime.strptime(end, "%Y-%m-%d")
    except ValueError:
        raise ValueError("Date format must be YYYY-MM-DD")

    cur = start_date
    list_date = []
    while cur<=end_date:
        list_date.append(cur)
        cur += timedelta(days=1)

    return np.array(list_date,dtype='datetime64[D]')


date_array(start = '2022-11-01', end = '2022-12-06')  

```

### `Q-18` Answer the following questions on the given array:

1. Fetch every alternate column of the array.
2. Normalize the given array.

Reference: [Normalization (Statistics)](https://en.wikipedia.org/wiki/Normalization_\(statistics\))

There are different forms of normalization. For this question, use the following formula:

$$
X_{normalized} = \frac{X - X_{min}}{X_{max} - X_{min}}
$$



What is Normalization?

Normalization is the process of converting values in a dataset to a common range, usually 0 to 1.

It is useful when different columns have very different scales.

```python
# Given
arr1=np.random.randint(low=1, high=10000, size=40).reshape(8,5)
arr1
```

```python
# Fetching every alternate col
arr1[:,::2] 
```

```python
# normalize the array 
def normalize(array):
    return (array-np.min(array))/(np.max(array)-np.min(array))


normalize(arr1)
```

Q-19: Write a function which will accept 2 arguments.
First: A 1D numpy array arr

Second: An integer n {Please make sure n<=len(arr)}

Output: The output should be the nth largest item out of the array

Example1 : arr=(12,34,40,7,1,0) and n=3, the output should be 12  
Example2 : arr=(12,34,40,7,1,0) and n=1, the output should be 40

```python
def Nth_largest(array,num):
    if array.ndim>1:
        raise TypeError("Requires 1d array")
        
    if num < 1 or num>array.size:
        raise ValueError("Nth Value should be less than and equal t0 array's length")

    return np.sort(array,descending=True)[num-1]

arr=np.random.randint(low=1,high=200,size=20) 
print(arr)
```

```python
Nth_largest(arr,4)
```

Q-20: Create the following pattern without hardcoding. Use only numpy functions and the below input array a.

| Function      | What it repeats     | Example               |
| ------------- | ------------------- | --------------------- |
| `np.repeat()` | Each **element**    | `[1,1,1,2,2,2,3,3,3]` |
| `np.tile()`   | The **whole array** | `[1,2,3,1,2,3,1,2,3]` |


```python
a = np.array([1,2,3])
new = np.repeat(a,np.size(a))
# np.concatenate((new,a,a,a))
# np.hstack([np.repeat(a, 3), np.tile(a, 3)])
np.concatenate((new, np.tile(a, np.size(a))))
```
